```python
def delta_solution(a, b, c):
    delta = (b**2) - (4*a*c)
    if delta > 0:
        x1 = (-b + delta ** (1/2)) / (2*a)
        x2 = (-b - delta ** (1/2)) / (2*a)
        return [x1, x2]
    elif delta == 0:
        x = (-b) / (2*a)
        return [x]
    else:
        return []

a = -1
b = -5
c = 6

delta_solution(a, b, c)
```
