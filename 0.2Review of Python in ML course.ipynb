```python
def countDog(st):
    words = st.lower().split()
    count = 0
    for word in words:
        if word == 'dog':
            count = count +1
    return count
```

```python
countDog('This dog runs faster than the other Dog')
```

```python
#abstract form
def countDog(st):
    count = 0
    for word in st.lower().split():
        if word == 'dog':
            count += 1
    return count
```

```python
countDog('This dog runs faster than the other Dog')
```

```python
countDog('There are four dogs in dog house at dog yard')
```

```python
def catch_speeding(speed, is_birthday):
    if is_birthday:
        speeding = speed - 5
    else:
        speeding = speed
    if speed >= 81:
        print('Big Ticket')
    elif speed == range(61, 80):
        print('Small Ticket')
    else:
        speed == range(0, 60)
        print('None Ticket')
```

```python
catch_speeding(81, True)
```

```python
catch_speeding(60, False)
```
