```python
#variable scope
x = 50
def func(x):
    print('x is ', x)
    x = 2
    print('Changed local x to ', x)
func(x)
print('x is still ', x)
```

```python
#global naming
x = 50

def func():
    global x
    print('This is function now using the global x!')
    print('Because of global x is: ', x, id(x))

    x = 2
    print('Run func(), changed global x to ', x, id(x))
print('Before calling func(), x is: ', x, id(x))
func()
print('Value of x (outside of func()) is: ', x, id(x))
```

```python
#Find even number in a list or range
def even_list(num_list):
    even_list = []
    for number in num_list:
        if number % 2 == 0:
            even_list.append(number)
            continue
            print(even_list)
        else:
            pass
    return even_list
```
