```python
#find prime number in list
def is_prime(num):
    '''
    Naive method of checking for primes
    '''
    for n in range(2, num):
        if num % n == 0:
            print(f"{num} is not a prime!")
            break
    else:
        print(f"{num} is a Prime Number!")

#this loop should work for all numbers in [2 to n]; so it needs a lot of sources. It's hard for big numbers.
#we can use mathematical principles to summarize it
```

```python
is_prime(10)
```

```python
is_prime(11)
```

```python
#1. every even number except 2 isn't prime. So, we can put even_numbers away.
#2. we can use square roots

import math
def is_prime(num):
    '''
    Better method of checking for primes.
    '''
    if num % 2 == 0 and num > 2:
        return False
    for i in range(3, int(math.sqrt(num)) + 1, 2):
        if num % i == 0:
            return False
    return True
```

```python
is_prime(5915587277)
```

```python
#find the lesser between two even numbers or the biger between even and odd number
def lesser(a, b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    else:
        return max(a, b)
```

```python
#don't consider 6 to 9 numbers in summation
def summer69(arr):
    total = 0
    add = True
    for num in arr:
        while add:
            if num != 6:
                total += num
                break
            else:
                add = False
        while not add:
            if num != 9:
                break
            else:
                add = True
                break
    return total

summer69([1, 3, 5])
```

```python
#find code as 0, 0 , 7 in list
def spy_game(nums):
    code = [0, 0, 7, 'x']
    for num in nums:
        if num == code[0]:
            code.pop(0)
    return len(code) == 1

spy_game([1, 2, 4, 0, 0, 7, 5])
```

```python
#find every prine number before the number and count them
def count_primes(num):
    primes = [2]
    x = 3
    if num < 2:
        return 0
    while x <= num:
        for y in range(3, x, 2):
            if x % y == 0:
                x += 2
                break
        else:
            primes.append(x)
            x += 2
    print(primes)
    return len(primes)

count_primes(100)
```

```python
#count the lower and upper case of each character in the text
def up_low(text):
    d = {'upper': 0, 'lower': 0}
    for char in text:
        if char.isupper():
            d['upper'] += 1
        elif char.islower():
            d['lower'] += 1
        else:
            pass
    print(f"Original string: {text}")
    print(f"No. of Upper case of characters: {d['upper']}")
    print(f"No. of Lower case of characters: {d['lower']}")

text = 'Hello, Mr. Rogers. How are you this Tuesday?'
up_low(text)
```

```python
#Is there at least once each english alphabet used in the text
import string
def is_pangram(str1, alphabet = string.ascii_lowercase):
    alphabet = set(alphabet)
    str1 = str1.replace(' ', '')
    str1 = str1.lower()
    str1 = set(str1)
    return str1 == alphabet

is_pangram('The quick brown fox jumps over the lazy dog')
```
